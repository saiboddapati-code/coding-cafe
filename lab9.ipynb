## Guided Practice: Working with NumPy Arrays

Let's start by creating a NumPy array containing the heights of five plants.

```python
growth = np.array([4,7,5,9,6,8])
```

We can now perform calculations directly on this array.

| Function | What it does |
|:---------|:-------------|
| `np.mean()` | Calculates the average of the values. |
| `np.max()` | Finds the largest value. |
| `np.min()` | Finds the smallest value. |
| `np.sum()` | Calculates the total of all values. |

Let's try these functions:

````python
import numpy as np
growth = np.array([4, 7, 5, 9, 6, 8])

average_growth = np.mean(growth)
maximum_growth = np.max(growth)
minimum_growth = np.min(growth)
total_growth = np.sum(growth)

print("Average growth:", average_growth)
print("Maximum growth:", maximum_growth)
print("Minimum growth:", minimum_growth)
print("Total growth:", total_growth)
````

### Task 1
Task: Temperature Monitoring

A scientist records the temperature inside a greenhouse over five days.

The recorded temperatures are:

`[22, 25, 21, 28, 24]`

Using NumPy:

- Create an array called `temperatures`.

- Calculate the average temperature.

- Find the highest and lowest temperatures.

- Find all temperatures above 23°C using Boolean filtering.

- Count how many days had temperatures above 23°C.

````python
# your code here
import numpy as np

# Step 1: Create the NumPy array
temperatures = np.array([22, 25, 21, 28, 24])

# Step 2: Calculate average temperature
average_temp = np.mean(temperatures)

# Step 3: Find highest and lowest temperatures
highest_temp = np.max(temperatures)
lowest_temp = np.min(temperatures)

# Step 4: Filter temperatures above 23°C
above_23 = temperatures[temperatures > 23]

# Step 5: Count days above 23°C
count_days = np.sum(temperatures > 23)

# Display results
print("Average temperature:", average_temp)
print("Highest temperature:", highest_temp)
print("Lowest temperature:", lowest_temp)
print("Temperatures above 23°C:", above_23)
print("Number of days above 23°C:", count_days)
````

### Selecting and Filtering Data

We can select a particular column by using its name.

We can also filter rows based on a condition.

For example, let's find all plants with a height greater than 12 cm.

````python
import pandas as pd

# Step 1: Create a dictionary
data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial_Height': [10, 12, 8, 11, 14],
    'Final_Height': [18, 20, 15, 17, 23],
    'Sunlight': [6, 7, 5, 8, 6]
}

# Step 2: Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Step 3: Display the complete DataFrame
print("Complete Plant Dataset:")
print(df)

# Step 4: Select the Plant column
print("\nPlant Names:")
print(df['Plant'])

# Step 5: Select the Final_Height column
print("\nFinal Heights:")
print(df['Final_Height'])

# Step 6: Filter plants with final height greater than 12 cm
tall_plants = df[df['Final_Height'] > 12]

print("\nPlants with height greater than 12 cm:")
print(tall_plants)

# Step 7: Filter plants with sunlight greater than 6 hours
sunny_plants = df[df['Sunlight'] > 6]

print("\nPlants receiving more than 6 hours of sunlight:")
print(sunny_plants)
````

### Grouping Data

The `groupby()` method allows us to group rows that share the same value.

For example, we can calculate the average height of each plant type.

````python
import pandas as pd

# Step 1: Create the dataset
data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial_Height': [10, 12, 8, 11, 14],
    'Final_Height': [18, 20, 15, 17, 23],
    'Sunlight': [6, 7, 5, 8, 6]
}

# Step 2: Create DataFrame
plants_df = pd.DataFrame(data)

# Step 3: Group plants by name and calculate average final height
average_height = plants_df.groupby('Plant')['Final_Height'].mean()

# Step 4: Display the result
print("Average height of each plant:")
print(average_height)
````

### Task 2
Continue working with the DataFrame created in the practice question.

- Create a new column called `Growth` by subtracting the initial height from the final height.

- Display only the `Plant` and `Growth` columns.

- Filter and display plants that grew more than 7 cm.

- Calculate the average growth for each plant type using `groupby()`.

````python
# your code here
import pandas as pd

# Step 1: Create the dictionary
data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial_Height': [10, 12, 8, 11, 14],
    'Final_Height': [18, 20, 15, 17, 23],
    'Sunlight': [6, 7, 5, 8, 6]
}

# Step 2: Create the DataFrame
df = pd.DataFrame(data)

# Step 3: Create a new column called Growth
df['Growth'] = df['Final_Height'] - df['Initial_Height']

# Step 4: Display only Plant and Growth columns
print("Plant and Growth:")
print(df[['Plant', 'Growth']])

# Step 5: Filter plants that grew more than 7 cm
high_growth = df[df['Growth'] > 7]

print("\nPlants that grew more than 7 cm:")
print(high_growth)

# Step 6: Calculate average growth for each plant type
average_growth = df.groupby('Plant')['Growth'].mean()

print("\nAverage growth for each plant type:")
print(average_growth)
````

### Commonly Used Matplotlib Functions

| Function | What it does |
|:---------|:-------------|
| `plt.bar()` | Creates a bar graph. |
| `plt.plot()` | Creates a line graph. |
| `plt.scatter()` | Creates a scatter plot. |
| `plt.hist()` | Creates a histogram. |
| `plt.title()` | Adds a title to the graph. |
| `plt.xlabel()` | Labels the x-axis. |
| `plt.ylabel()` | Labels the y-axis. |
| `plt.grid()` | Adds grid lines. |

Let's try a line graph using the same plant growth data.

````python
import matplotlib.pyplot as plt

# Step 1: Create a list of plant names
plants = ['Rose', 'Tulip', 'Lily', 'Sunflower']

# Step 2: Create a list of plant heights
heights = [15, 20, 12, 25]

# Step 3: Create a bar graph
plt.bar(plants, heights)

# Step 4: Add title and axis labels
plt.title('Comparison of Plant Heights')
plt.xlabel('Plant Names')
plt.ylabel('Height (cm)')

# Step 5: Display the bar graph
plt.show()


# ---------------------------------
# Line Graph using the same data
# ---------------------------------

# Step 6: Create a line graph
plt.plot(plants, heights, marker='o')

# Step 7: Add title and axis labels
plt.title('Plant Height Comparison - Line Graph')
plt.xlabel('Plant Names')
plt.ylabel('Height (cm)')

# Step 8: Add grid lines
plt.grid(True)

# Step 9: Display the line graph
plt.show()
````

## Part 4 : Combining everything!
### The School Science Fair

Your school has just held its annual Science Fair! Six students participated, and their marks in Physics, Chemistry and Biology have been recorded.

Your job is to analyse their performance and identify the top-performing students.

### Research Dataset

| Student | Physics | Chemistry | Biology |
|:--------|--------:|-----------:|--------:|
| Asha | 85 | 78 | 92 |
| Jess | 72 | 88 | 80 |
| Meera | 95 | 91 | 89 |
| Rohan | 65 | 70 | 75 |
| Sara | 88 | 84 | 90 |
| Dev | 78 | 76 | 82 |

### Your Mission

**Step 1: Create the Dataset using Pandas**

- Store the information in a dictionary.
- Convert the dictionary into a DataFrame called `students_df`.
- Display the dataset.

**Step 2: Analyse the Scores using NumPy**

- Convert the Physics scores into a NumPy array.
- Calculate the average Physics score.
- Find the highest and lowest Physics scores.

**Step 3: Analyse Student Performance using Pandas**

- Create a new column called `Total` by adding the three subject scores.
- Create another column called `Average` to calculate each student's average marks.
- Filter and display students whose average score is greater than 80.
- Find the student with the highest average score.

**Step 4: Visualise Your Findings using Matplotlib**

- Create a bar graph comparing the average scores of all six students.
- Add a suitable title and label both axes.
- Add a grid to make the graph easier to read.

**Step 5: Answer the Following Questions**

1. Which student scored the highest in Physics?
2. What was the average Physics score?
3. Which student had the highest overall average?
4. How many students scored an average above 80?

````python
# Step 1: Create the Dataset using Pandas

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Store student information in a dictionary
data = {
    'Student': ['Asha', 'Jess', 'Meera', 'Rohan', 'Sara', 'Dev'],
    'Physics': [85, 72, 95, 65, 88, 78],
    'Chemistry': [78, 88, 91, 70, 84, 76],
    'Biology': [92, 80, 89, 75, 90, 82]
}

# Convert dictionary into a DataFrame
students_df = pd.DataFrame(data)

# Display the dataset
print("STUDENT DATASET:")
print(students_df)


# Step 2: Analyse the Scores using NumPy

# Convert Physics scores into a NumPy array
physics_scores = np.array(students_df['Physics'])

# Calculate average, highest and lowest Physics scores
average_physics = np.mean(physics_scores)
highest_physics = np.max(physics_scores)
lowest_physics = np.min(physics_scores)

print("\nPHYSICS ANALYSIS:")
print("Average Physics Score:", average_physics)
print("Highest Physics Score:", highest_physics)
print("Lowest Physics Score:", lowest_physics)


# Step 3: Analyse Student Performance using Pandas

# Calculate Total marks
students_df['Total'] = (
    students_df['Physics'] +
    students_df['Chemistry'] +
    students_df['Biology']
)

# Calculate Average marks
students_df['Average'] = students_df['Total'] / 3

print("\nSTUDENT PERFORMANCE:")
print(students_df)

# Filter students whose average is greater than 80
above_80 = students_df[students_df['Average'] > 80]

print("\nSTUDENTS WITH AVERAGE ABOVE 80:")
print(above_80)

# Find the student with the highest average
top_student = students_df.loc[students_df['Average'].idxmax()]

print("\nTOP-PERFORMING STUDENT:")
print(top_student)


# Step 4: Visualise Findings using Matplotlib

plt.bar(students_df['Student'], students_df['Average'])

plt.title('Average Scores of Students')
plt.xlabel('Students')
plt.ylabel('Average Marks')

plt.grid(axis='y')
plt.show()


# Step 5: Answer the Questions

print("\nFINAL ANSWERS:")

# Student with highest Physics score
physics_topper = students_df.loc[students_df['Physics'].idxmax(), 'Student']
print("1. Highest Physics scorer:", physics_topper)

# Average Physics score
print("2. Average Physics score:", average_physics)

# Student with highest overall average
print("3. Highest overall average:", top_student['Student'])

# Number of students with average above 80
print("4. Number of students scoring above 80:", len(above_80))
````
